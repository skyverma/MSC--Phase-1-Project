```python
from transformers import AutoTokenizer
import csv

# Initialize the tokenizer
tokenizer = AutoTokenizer.from_pretrained("t5-small")
max_token_id = tokenizer.vocab_size - 1  # Maximum valid token ID for the tokenizer

# Input and output file paths
input_file_path = "/content/submission_epoch_1.csv"  # Replace with your input file path
output_file_path = "/content/decoded_sentences.csv"  # Path for the output CSV

# Initialize lists for successful and failed rows
successful_rows = []
failed_rows = []

# Step 1: Process the CSV file
with open(input_file_path, "r") as infile:
    reader = csv.DictReader(infile)  # Read the input as a dictionary
    for row in reader:
        row_id = row.get("ID", "Unknown")  # Get the ID column if it exists
        text_codes = row.get("Text", "")  # Get the Text column

        try:
            # Parse token IDs (assuming they're space-separated integers)
            token_ids = list(map(int, text_codes.strip().split()))
            # Validate token IDs (filter out invalid ones)
            token_ids = [tid for tid in token_ids if 0 <= tid <= max_token_id]
            # Decode the token IDs into a sentence
            decoded_text = tokenizer.decode(token_ids, skip_special_tokens=True)
            # Append successful row
            successful_rows.append({"ID": row_id, "Text": decoded_text})
        except (ValueError, OverflowError) as e:
            # Log failed rows for debugging
            failed_rows.append({"ID": row_id, "Error": str(e)})

# Step 2: Write the successful rows to an output CSV
with open(output_file_path, "w", newline="") as outfile:
    fieldnames = ["ID", "Text"]
    writer = csv.DictWriter(outfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(successful_rows)

# Step 3: Log problematic rows for review
if failed_rows:
    error_log_path = "/content/failed_rows.log"
    with open(error_log_path, "w") as error_file:
        for failed_row in failed_rows:
            error_file.write(f"Row ID: {failed_row['ID']}, Error: {failed_row['Error']}\n")
    print(f"Problematic rows logged to {error_log_path}.")

print(f"Decoded sentences saved to {output_file_path}.")

```
